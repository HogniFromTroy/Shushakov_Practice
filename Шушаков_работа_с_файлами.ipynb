## Задача 1. Создать файл и прочитать

*Новое: `open`, режимы `'w'` и `'r'`, `write`, `read`*

Режим `'w'` создаёт файл (или стирает старый), `write()` записывает строку.
Перенос строки `\n` нужно ставить самому. Режим `'r'` открывает файл на чтение,
`read()` возвращает всё содержимое одной строкой.

````python
# создаём файл и записываем две строки
with open('note.txt', 'w', encoding='utf-8') as file:
    file.write('Купить хлеб\n')
    file.write('Позвонить маме\n')

# открываем и читаем целиком
with open('note.txt', 'r', encoding='utf-8') as file:
    text = file.read()

print(text)
print('символов в файле:', len(text))
````

## Задача 2. Список в файл, файл по строкам

*Новое: запись циклом, `for line in file`, `strip`*

Список удобно записать **циклом**, где каждый элемент - это отдельная строка с `\n` в конце строки.
Читать файл тоже удобно циклом `for line in file`, где в `line` по очереди попадает каждая
строка вместе с `\n`, поэтому её очищают методом `strip()`. Номера строк выводят при помощи функции `enumerate`.

````python
shopping = ['хлеб', 'молоко', 'сыр', 'яблоки']

with open('shopping.txt', 'w', encoding='utf-8') as file:
    for item in shopping:
        file.write(item + '\n')

with open('shopping.txt', 'r', encoding='utf-8') as file:
    for number, line in enumerate(file, start=1):
        print(f'{number}. {line.strip()}')
````

## Задача 4. Строки в обратном порядке

*Новое: `[::-1]`, `writelines`, новый файл*

Читаем все строки списком, переворачиваем
срезом `[::-1]` и записываем в **новый** файл. Строки из `readlines()` уже заканчиваются
на `\n`, поэтому их можно записать как есть при помощи метода `writelines()`.

````python
poem = ['Roses are red,', 'Violets are blue,', 'Sugar is sweet,', 'And so are you.']

with open('poem.txt', 'w', encoding='utf-8') as file:
    for line in poem:
        file.write(line + '\n')

with open('poem.txt', 'r', encoding='utf-8') as file:
    lines = file.readlines()

with open('reversed_poem.txt', 'w', encoding='utf-8') as file:
    file.writelines(lines[::-1])

with open('reversed_poem.txt', 'r', encoding='utf-8') as file:
    print(file.read())
````

## Задача 5. Склеить содержимое нескольких файлов

*Новое: список имён файлов, `'---\n'.join`*

Имена файлов храним в списке и
обходим циклом. Тексты собираем в список, а склеиваем методом `join`: разделитель
`'---\n'` встанет **между** текстами, но не после последнего.

````python
parts = ['The quick brown fox', 'jumps over the lazy dog', 'and runs away swiftly.']
names = ['file1.txt', 'file2.txt', 'file3.txt']

# создаём три файла
for i in range(len(names)):
    with open(names[i], 'w', encoding='utf-8') as file:
        file.write(parts[i] + '\n')

# читаем каждый и склеиваем
texts = []
for name in names:
    with open(name, 'r', encoding='utf-8') as file:
        texts.append(file.read())

with open('combined.txt', 'w', encoding='utf-8') as file:
    file.write('---\n'.join(texts))

with open('combined.txt', 'r', encoding='utf-8') as file:
    print(file.read())
````

## Задача 6. Подсчёт слов и частоты

*Новое: слова из файла, словарь частот*

Текст из файла разделяем
на слова, приводим к нижнему регистру и убираем знаки препинания по краям,
иначе `language` и `language.` будут разными словами. Частоты считаем словарём
и сортируем по убыванию (`key=` **из темы 2**).

Самые частые слова почти всегда короткие служебные: `is`, `and`, `a`. Поэтому слова
короче 4 букв часто просто не учитывают.

````python
sample = [
    'Natural Language Processing, or NLP, is a field of artificial intelligence that focuses on the interaction between computers and humans through natural language.',
    'The ultimate objective of NLP is to enable computers to understand, interpret, and generate human languages.',
    'NLP is a critical technology in many fields, including translation, sentiment analysis, and speech recognition.',
]
with open('text_sample.txt', 'w', encoding='utf-8') as file:
    for line in sample:
        file.write(line + '\n')

with open('text_sample.txt', 'r', encoding='utf-8') as file:
    text = file.read()

words = [word.strip('.,:%').lower() for word in text.split()]
print('всего слов:', len(words))
print('разных слов:', len(set(words)))

freq = {}
for word in words:
    freq[word] = freq.get(word, 0) + 1

top = sorted(freq.items(), key=lambda pair: pair[1], reverse=True)
print('чаще всего:', top[:3])

long_words = [word for word in words if len(word) >= 4]
long_freq = {}
for word in long_words:
    long_freq[word] = long_freq.get(word, 0) + 1
long_top = sorted(long_freq.items(), key=lambda pair: pair[1], reverse=True)
print('чаще всего от 4 букв:', long_top[:3])
````

## Задача 7. Отбор строк в новый файл

````python
logs = [
    'INFO: User logged in',
    'ERROR: Failed to connect to database',
    'INFO: Data processed successfully',
    'ERROR: Timeout while fetching data',
    'INFO: User logged out',
]
with open('logs.txt', 'w', encoding='utf-8') as file:
    for line in logs:
        file.write(line + '\n')

with open('logs.txt', 'r', encoding='utf-8') as file:
    lines = file.readlines()

with open('errors.txt', 'w', encoding='utf-8') as file:
    for line in lines:
        if line.startswith('ERROR'):
            file.write(line)

with open('errors.txt', 'r', encoding='utf-8') as file:
    print(file.read())
````

## Задача 8. Итоги по CSV

*Новое: CSV через `split(',')`, `int`, словарь итогов*

CSV - это обычный текстовый файл, где первая строка - это
заголовок, а значения разделены запятыми. Разбираем каждую строку через `split(',')`.
Всё, что прочитано из файла - это строки, поэтому числа преобразуем в `int`.
Итоги копим в словаре и записываем в новый CSV-файл.

````python
with open('sales.csv', 'w', encoding='utf-8') as file:
    file.write('product,amount\n')
    file.write('apple,5\n')
    file.write('banana,3\n')
    file.write('apple,7\n')
    file.write('orange,2\n')
    file.write('banana,4\n')

totals = {}
with open('sales.csv', 'r', encoding='utf-8') as file:
    file.readline()                               # пропускаем заголовок
    for line in file:
        product, amount = line.strip().split(',')
        totals[product] = totals.get(product, 0) + int(amount)

with open('summary.csv', 'w', encoding='utf-8') as file:
    file.write('product,total_amount\n')
    for product in totals:
        file.write(f'{product},{totals[product]}\n')

with open('summary.csv', 'r', encoding='utf-8') as file:
    print(file.read())
````

## Шпаргалка занятия

| Что нужно | Как пишется |
| --- | --- |
| создать / перезаписать | `with open('f.txt', 'w', encoding='utf-8') as file:` |
| дописать в конец | `with open('f.txt', 'a', encoding='utf-8') as file:` |
| прочитать | `with open('f.txt', 'r', encoding='utf-8') as file:` |
| записать строку | `file.write(text + '\n')` |
| записать число | `file.write(str(n) + '\n')` или `file.write(f'{n}\n')` |
| прочитать всё | `text = file.read()` |
| список строк | `lines = file.readlines()` |
| перебрать строки | `for line in file:` и `line.strip()` |
| строки наоборот | `file.writelines(lines[::-1])` |
| склеить через разделитель | `'---\n'.join(texts)` |
| отобрать строки | `if line.startswith('ERROR'): file.write(line)` |
| строка CSV | `a, b = line.strip().split(',')`, числа - `int(b)` |
| пропустить заголовок | `file.readline()` перед циклом |
